```python
clarity_order={'IF' : 1,'VVS1' : 2,'VVS2' :3 ,'VS1': 4 ,'VS2' :5,'SI1':6 , 'SI2':7,
'I1': 8}
df['clarity_encode'] = df['clarity'].map(clarity_order)
```

```python
df=df.drop(columns="clarity")
```

```python
df['clarity_encode'].info()
```

```python
x_one_encode = pd.get_dummies(df,columns=['cut' , 'color'],drop_first= True)
```

```python
x_one_encode.head()
```

```python
x_one_encode=x_one_encode.astype(int)
```

```python
x_one_encode.head()
```

```python
X = x_one_encode.drop(columns=['price'])
Y = x_one_encode['price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train , Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python

from sklearn.metrics import mean_squared_error, r2_score
```

```python

print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
