```python
xencode = pd.get_dummies(df,columns=['Gender','education','prevalentStroke'], drop_first= True)
```

```python
xencode.head()
```

```python
bool_col= xencode.select_dtypes(include='bool').columns
xencode[bool_col]= xencode[bool_col].astype(int) 
```

```python
xencode['Heart_ stroke'] = xencode['Heart_ stroke'].map({'No':0,'yes':1})
```

```python
xencode.head()
```

```python
xencode.info()
```

```python
xencode = xencode.drop(columns=['smoke_level','chol_level','sysBP_levels','diaBP_levels','BMI_levels','heartRate_levels','glucose_levels'])
```

```python
xencode.head()
```

```python
x= xencode.drop(columns=['Heart_ stroke'])
y= xencode['Heart_ stroke']
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x_train , x_test , y_train , y_test =train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
scaler= StandardScaler()
x_train_scale= scaler.fit_transform(x_train)
x_test_scale= scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scale,y_train)
```

```python
y_pred= knn.predict(x_test_scale)
```

```python
from sklearn.metrics import accuracy_score, classification_report , confusion_matrix
```

```python
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

yeee we got 83% accuracy score and further , but their is problem as we can see TN score in confusion matrix is too low , i figure it out the main reason why this problem occur actually due to imbalance in data which has 3256 'No' and only 593 'yes' (about 85% vs 15%).

KNN uses majority votes thats why 'No' has majority votes among them.

**USING LOGISTC REGRESSION**

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train_scale,y_train)
```

```python
y_pre2= model.predict(x_test_scale)
```

```python
print(accuracy_score(y_test,y_pre2))
print(confusion_matrix(y_test,y_pre2))
print(classification_report(y_test,y_pre2))
```
